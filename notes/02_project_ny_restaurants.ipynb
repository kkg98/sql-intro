## Task 11. New rating system with `CASE`

**Mistake: commas between the `WHEN` lines.** I first wrote `then 'Extraordinary',` and the query failed.
A comma separates **columns** in `SELECT`, so SQL read `'Extraordinary',` as the end of the column and choked on the next `when`.
`WHEN` lines inside one `CASE` have no separator. Fixed below.

**Watch the output:** Jacob's Pickles and Smalls Jazz Club get "Poor", but they have **no** review (`NULL`).
`NULL` fails every `WHEN`, so it lands in `ELSE`. A `when review is null then 'No reviews'` line first would fix that.

```python
%%sql
select name,
case
  when review > 4.5 then 'Extraordinary'
  when review > 4 then 'Excellent'
  when review > 3 then 'Good'
  when review > 2 then 'Fair'
  else 'Poor'
end as 'Food is...'
from restaurants;
```
